# IAN 601 — Module 4: Lists & Tuples

Lists and tuples let Python organize collections of related values.

## 1. Create a list

```python
scores = [82, 76, 91, 88]

print(scores)
print(type(scores))
```

## 2. Index and slice a list

```python
scores = [82, 76, 91, 88, 95]

print("First:", scores[0])
print("Last:", scores[-1])
print("First three:", scores[:3])
print("Middle:", scores[1:4])
```

## 3. Change a list

```python
scores = [82, 76, 91]

scores.append(88)
scores[1] = 80

print(scores)
```

## 4. Loop through a list

```python
scores = [82, 76, 91, 88]

for score in scores:
    print(score)
```

## 5. Summarize values in a list

```python
scores = [82, 76, 91, 88]

print("Count:", len(scores))
print("Total:", sum(scores))
print("Average:", sum(scores) / len(scores))
print("Minimum:", min(scores))
print("Maximum:", max(scores))
```

## 6. Create a tuple

A tuple is useful when the values belong together and should not be changed accidentally.

```python
patient = ("P104", 42, "Greensboro")

print(patient)
print(patient[0])
```

## 7. Unpack a tuple

```python
patient = ("P104", 42, "Greensboro")

patient_id, age, city = patient

print(patient_id)
print(age)
print(city)
```

## 8. Lists can hold tuples

```python
patients = [
    ("P101", 35),
    ("P102", 52),
    ("P103", 28)
]

for patient_id, age in patients:
    print(patient_id, age)
```

## 9. Patient-age list management

```python
ages = [35, 52, 28, 44, 61]

ages.append(39)
ages.remove(28)

adult_ages = [age for age in ages if age >= 18]

print("Ages:", ages)
print("Average age:", round(sum(ages) / len(ages), 1))
print("Adults:", adult_ages)
```

## 10. Choose the right sequence

- Use a list when values may be added, removed, or changed.
- Use a tuple when a small record should stay fixed.
- A list can contain tuples when you need many fixed records.

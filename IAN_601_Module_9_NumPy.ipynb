Think in arrays: calculate, filter, reshape, and summarize many numeric values at once.

## 1. Python lists vs NumPy arrays

```python
import numpy as np

python_list = [10, 20, 30]
array_values = np.array([10, 20, 30])

print("List + list:", python_list + python_list)
print("Array + array:", array_values + array_values)
print("Array * 2:", array_values * 2)
```

## 2. Create your first array

```python
scores = np.array([88, 92, 79, 95, 84])

print(scores)
print(type(scores))
print("Mean:", scores.mean())
```

## 3. Useful creation functions

```python
days = np.arange(1, 8)
thresholds = np.linspace(0, 1, 5)
zeros = np.zeros((2, 3))
ones = np.ones(4)

print("Days:", days)
print("Thresholds:", thresholds)
print("Zeros:\n", zeros)
print("Ones:", ones)
```

## 4. dtype, shape, ndim, and size

```python
sales = np.array([
    [120, 135, 150],
    [80, 92, 88]
], dtype=float)

print("dtype:", sales.dtype)
print("shape:", sales.shape)
print("ndim:", sales.ndim)
print("size:", sales.size)
```

## 5. Vectorized calculations

```python
grades = np.array([74, 81, 89, 93, 68])

curved = np.minimum(grades + 5, 100)

print("Original:", grades)
print("Curved:", curved)
```

## 6. Universal functions

```python
values = np.array([1, 4, 9, 16], dtype=float)

print("Square root:", np.sqrt(values))
print("Natural log:", np.log(values).round(3))
print("Absolute:", np.abs(np.array([-3, 2, -7])))
```

## 7. Index and slice 1D arrays

```python
scores = np.array([88, 92, 79, 95, 84, 91])

print("First:", scores[0])
print("Last:", scores[-1])
print("First three:", scores[:3])
print("Every other:", scores[::2])
```

## 8. Boolean masks

```python
scores = np.array([88, 92, 79, 95, 84, 91])

mask = scores >= 90

print(mask)
print(scores[mask])
```

## 9. np.where()

```python
scores = np.array([88, 92, 79, 95, 84])

labels = np.where(scores >= 90, "high", "standard")

print(labels)
```

## 10. Two-dimensional arrays

```python
sales = np.array([
    [120, 135, 150],
    [80, 92, 88],
    [200, 180, 210]
])

print(sales)
print("Row 0:", sales[0])
print("Column 1:", sales[:, 1])
print("Single value:", sales[2, 2])
```

## 11. Reshape arrays

```python
values = np.arange(1, 13)

matrix = values.reshape(3, 4)

print(matrix)
print("Shape:", matrix.shape)
```

## 12. Transpose

```python
matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Original:\n", matrix)
print("Transpose:\n", matrix.T)
```

## 13. Summaries and axes

```python
sales = np.array([
    [120, 135, 150],
    [80, 92, 88],
    [200, 180, 210]
])

print("Overall mean:", sales.mean())
print("Column means:", sales.mean(axis=0))
print("Row totals:", sales.sum(axis=1))
```

## 14. Broadcasting

```python
sales = np.array([
    [100, 120, 140],
    [80, 90, 110]
])

monthly_adjustment = np.array([1.00, 1.05, 1.10])

adjusted = sales * monthly_adjustment

print(adjusted)
```

## 15. Concatenate arrays

```python
a = np.array([10, 20, 30])
b = np.array([40, 50])

combined = np.concatenate([a, b])

print(combined)
```

## 16. Missing numeric values with NaN

```python
values = np.array([10.0, 12.5, np.nan, 18.0])

print("Mean:", np.mean(values))
print("NaN-aware mean:", np.nanmean(values))
print("Missing count:", np.isnan(values).sum())
```

## 17. Percentiles and spread

```python
values = np.array([45, 52, 61, 72, 88, 91, 105])

print("Median:", np.median(values))
print("25th percentile:", np.percentile(values, 25))
print("75th percentile:", np.percentile(values, 75))
print("Std dev:", np.std(values).round(2))
```

## 18. Reproducible random generation

```python
rng = np.random.default_rng(601)

sample = rng.normal(
    loc=100,
    scale=15,
    size=10
)

print(sample.round(2))
```

## 19. Normalize numeric values

```python
values = np.array([40, 55, 70, 85, 100], dtype=float)

mean = values.mean()
std = values.std()

z_scores = (values - mean) / std

print(z_scores.round(3))
```

## 20. Matrix-style calculation

```python
features = np.array([
    [1.0, 2.0],
    [2.0, 1.5],
    [3.0, 4.0]
])

weights = np.array([0.8, 1.2])

scores = features @ weights

print(scores)
```

## 21. Copies versus views

```python
original = np.array([10, 20, 30, 40])

view = original[1:3]
copy = original[1:3].copy()

view[0] = 999

print("Original:", original)
print("View:", view)
print("Copy:", copy)
```

## 22. Validate assumptions

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

assert data.ndim == 2
assert data.shape[1] == 3
assert not np.isnan(data).any()

print("Array checks passed")
```

## 23. Mini NumPy workflow

```python
rng = np.random.default_rng(601)

sales = rng.integers(50, 250, size=30)

high_sales = sales[sales >= 150]

print("Shape:", sales.shape)
print("Average:", sales.mean().round(2))
print("High-sales count:", len(high_sales))
print("Top 5:", np.sort(sales)[-5:])
```

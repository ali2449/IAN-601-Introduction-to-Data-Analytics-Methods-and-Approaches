## 1. Readable code starts with names

```python
monthly_sales = [1200, 1350, 1280, 1410]
month_count = len(monthly_sales)
average_monthly_sales = sum(monthly_sales) / month_count

print("Average monthly sales:", round(average_monthly_sales, 2))
```

## 2. Comments explain why

```python
orders = [
    {"id": 101, "status": "complete", "amount": 45},
    {"id": 102, "status": "cancelled", "amount": 0},
    {"id": 103, "status": "complete", "amount": 72},
]

# Cancelled orders are excluded because they did not produce revenue.
completed_orders = [
    order for order in orders
    if order["status"] == "complete"
]

revenue = sum(order["amount"] for order in completed_orders)

print("Revenue:", revenue)
```

## 3. Exceptions help programs fail safely

```python
def to_number(value):
    try:
        return float(value)
    except (TypeError, ValueError):
        return None

raw_values = ["42.5", "N/A", "", None, "18"]
clean_values = [to_number(value) for value in raw_values]

print(clean_values)
```

## 4. Regular expressions validate text patterns

```python
import re

pattern = r"^[A-Z]{2}-\d{4}$"
ids = ["AB-1204", "A-1204", "xy-9999", "NC-2026"]

for value in ids:
    print(value, "->", bool(re.match(pattern, value)))
```

## 5. Extract useful text with regex

```python
import re

support_note = "Contact maria@example.com or sales@shop.org for help."

email_pattern = r"[\w\.-]+@[\w\.-]+\.\w+"

emails = re.findall(email_pattern, support_note)

print(emails)
```

## 6. Docstrings explain reusable code

```python
def conversion_rate(visits, signups):
    """Return the signup conversion rate as a percentage."""
    if visits == 0:
        return 0
    return (signups / visits) * 100

print(round(conversion_rate(250, 37), 2))
help(conversion_rate)
```

## 7. Type hints make intent visible

```python
def average(values: list[float]) -> float:
    if not values:
        return 0.0
    return sum(values) / len(values)

scores = [88.0, 92.5, 79.0]

print(average(scores))
```

## 8. Testing with assert

```python
def normalize_status(value):
    return value.strip().lower()

assert normalize_status(" Complete ") == "complete"
assert normalize_status("CANCELLED") == "cancelled"
assert normalize_status(" pending") == "pending"

print("All checks passed")
```

## 9. A small test suite

```python
def normalize_status(value):
    return value.strip().lower().replace(" ", "_")

def test_complete_status():
    assert normalize_status(" Complete ") == "complete"

def test_in_progress_status():
    assert normalize_status("In Progress") == "in_progress"

for test in [test_complete_status, test_in_progress_status]:
    test()

print("All tests passed")
```

## 11. Responsible data handling

```python
def mask_email(email):
    username, domain = email.split("@")
    return username[0] + "***@" + domain

emails = [
    "maria@example.com",
    "jordan@shop.org"
]

masked = [mask_email(email) for email in emails]

print(masked)
```

## 13. Mini project: clean order records

```python
def to_number(value):
    try:
        return float(value)
    except (TypeError, ValueError):
        return None

def normalize_status(value):
    return value.strip().lower()

orders = [
    {"id": 101, "amount": "45.50", "status": " Complete "},
    {"id": 102, "amount": "N/A", "status": "cancelled"},
    {"id": 103, "amount": "72.00", "status": " COMPLETE "}
]

clean_orders = []

for order in orders:
    amount = to_number(order["amount"])
    status = normalize_status(order["status"])

    if amount is None:
        continue

    clean_orders.append({
        "id": order["id"],
        "amount": amount,
        "status": status
    })

clean_orders
```

## 14. Summarize and test

```python
completed_revenue = sum(
    order["amount"]
    for order in clean_orders
    if order["status"] == "complete"
)

assert len(clean_orders) == 2
assert completed_revenue == 117.5

print("Completed revenue:", completed_revenue)
print("Tests passed")
```

## 1. Conditions ask True/False questions

```python
sales = 12000
target = 10000

print(sales > target)
print(sales == target)
print(sales != target)
```

## 2. if statements

```python
sales = 12000
target = 10000

if sales >= target:
    print("Target met")
```

## 3. if / else

```python
score = 78

if score >= 70:
    print("Pass")
else:
    print("Review needed")
```

## 4. elif handles several categories

```python
sales = 8200

if sales >= 10000:
    level = "high"
elif sales >= 5000:
    level = "medium"
else:
    level = "low"

print(level)
```

## 5. Boolean logic

```python
amount = 140
status = "complete"

flag = amount >= 100 and status == "complete"

print(flag)
```

## 6. for loops repeat over a sequence

```python
sales = [1200, 1350, 1280, 1410]

for value in sales:
    print(value)
```

## 7. range()

```python
for day in range(1, 6):
    print("Day", day)
```

## 8. Running totals

```python
orders = [45, 72, 38, 91]

total = 0

for amount in orders:
    total += amount

print("Total:", total)
```

## 9. enumerate()

```python
channels = ["web", "email", "store"]

for position, channel in enumerate(channels, start=1):
    print(position, channel)
```

## 10. while loops repeat until something changes

```python
balance = 100
payment = 25

while balance > 0:
    balance -= payment
    print("Balance:", max(balance, 0))
```

## 11. break and continue

```python
values = [12, -1, 18, 0, 25]

for value in values:
    if value < 0:
        continue
    if value == 0:
        break
    print(value)
```

## 12. Combine loops and decisions

```python
sales = [420, 1200, 860, 1500, 300]

for value in sales:
    if value >= 1000:
        label = "high"
    elif value >= 500:
        label = "medium"
    else:
        label = "low"

    print(value, "->", label)
```

## 13. Count values that meet a condition

```python
amounts = [45, 72, 138, 91, 165, 40]

large_count = 0

for amount in amounts:
    if amount >= 100:
        large_count += 1

print("Large orders:", large_count)
```

## 14. Final activity: customer flags

```python
customers = [
    {"id": "C1", "visits": 8, "spend": 420},
    {"id": "C2", "visits": 2, "spend": 90},
    {"id": "C3", "visits": 6, "spend": 280},
]

for customer in customers:
    if customer["visits"] >= 5 and customer["spend"] >= 250:
        flag = "priority"
    else:
        flag = "standard"

    print(customer["id"], flag)
```

## 5. What is a package?

A package is a folder that groups related Python modules. Libraries such as NumPy and Pandas are larger collections of reusable code.

## 6. Think like a small module

Keep related functions together and give them clear names.

```python
def clean_status(value):
    return value.strip().lower()

def is_complete(value):
    return clean_status(value) == "complete"

def valid_amount(value):
    return value is not None and value >= 0

print(clean_status(" Complete "))
print(is_complete(" COMPLETE "))
print(valid_amount(45))
```

## 7. What is a class?

A class is a blueprint for creating objects that combine data and behavior.

```python
class Customer:
    def __init__(self, customer_id, visits):
        self.customer_id = customer_id
        self.visits = visits

customer = Customer("C104", 7)

print(customer.customer_id)
print(customer.visits)
```

## 8. One class can create many objects

```python
class Customer:
    def __init__(self, customer_id, visits):
        self.customer_id = customer_id
        self.visits = visits

c1 = Customer("C101", 4)
c2 = Customer("C102", 9)

print(c1.customer_id, c1.visits)
print(c2.customer_id, c2.visits)
```

## 9. Methods and self

```python
class Customer:
    def __init__(self, customer_id, visits):
        self.customer_id = customer_id
        self.visits = visits

    def is_frequent(self):
        return self.visits >= 5

customer = Customer("C104", 7)

print(customer.is_frequent())
```

## 10. Small analytics class

```python
class SalesSummary:
    def __init__(self, revenue, costs):
        self.revenue = revenue
        self.costs = costs

    def profit(self):
        return self.revenue - self.costs

    def margin(self):
        if self.revenue == 0:
            return 0
        return self.profit() / self.revenue

summary = SalesSummary(18000, 12500)

print("Profit:", summary.profit())
print("Margin:", f"{summary.margin():.1%}")
```

## 11. Function, dictionary, or class?

- Function: reusable action or calculation.
- Dictionary: one flexible labeled record.
- Class: repeated objects that share data and behavior.
- Module: reusable Python file containing related code.

## 1. Variables store values

```python
customer_name = "Avery"
monthly_sales = 12500
tax_rate = 0.07

print(customer_name)
print(monthly_sales)
print(tax_rate)
```

## 2. Variables can be reassigned

```python
monthly_sales = 12500
print("Before:", monthly_sales)

monthly_sales = 13200
print("After:", monthly_sales)
```

## 3. Assign several values at once

```python
revenue, costs, customers = 18000, 12500, 240

print(revenue, costs, customers)
```

## 4. Data types

```python
order_id = "00125"
quantity = 3
price = 19.95
is_complete = True

print(type(order_id))
print(type(quantity))
print(type(price))
print(type(is_complete))
```

Some numeric-looking values should stay as text when arithmetic does not make sense, such as IDs, ZIP codes, and phone numbers.

## 5. Arithmetic creates derived metrics

```python
revenue = 18500
costs = 13200

profit = revenue - costs
margin = profit / revenue

print("Profit:", profit)
print("Margin:", margin)
```

## 6. Useful operators and parentheses

```python
sales = 2500
tax_rate = 0.07
discount = 100

total = (sales - discount) * (1 + tax_rate)

print(total)
```

## 7. f-strings create readable messages

```python
revenue = 18500
profit = 5300
margin = profit / revenue

print(f"Revenue: ${revenue:,.2f}")
print(f"Profit: ${profit:,.2f}")
print(f"Margin: {margin:.1%}")
```

## 8. Input usually starts as text

```python
raw_quantity = "4"
raw_price = "22.50"

quantity = int(raw_quantity)
price = float(raw_price)

total = quantity * price

print("Total:", total)
```

## 9. Boolean conversion previews decision logic

```python
sales = 12000
target = 10000

met_target = sales >= target

print("Met target?", met_target)
```

## 10. Preview a conversion error

```python
raw_value = "12.5"

number = float(raw_value)
print(number)

# Try changing raw_value to "N/A" and run again.
```

## 11. Mini project: simple sales calculator

```python
product_name = "Analytics Workbook"
quantity = 8
unit_price = 24.95
tax_rate = 0.07

subtotal = quantity * unit_price
tax = subtotal * tax_rate
total = subtotal + tax

print(f"Product: {product_name}")
print(f"Subtotal: ${subtotal:,.2f}")
print(f"Tax: ${tax:,.2f}")
print(f"Total: ${total:,.2f}")
```

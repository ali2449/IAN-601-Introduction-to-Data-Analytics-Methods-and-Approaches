# IAN 601 — Module 5: Dictionaries & Sets

Dictionaries store labeled values. Sets store unique values.

## 1. Create a dictionary

```python
customer = {
    "customer_id": "C104",
    "visits": 7,
    "status": "active"
}

print(customer)
```

## 2. Access values safely

```python
customer = {
    "customer_id": "C104",
    "visits": 7,
    "status": "active"
}

print(customer["customer_id"])
print(customer.get("visits"))
print(customer.get("segment", "unknown"))
```

## 3. Update a dictionary

```python
customer = {
    "customer_id": "C104",
    "visits": 7
}

customer["visits"] = 8
customer["status"] = "active"

print(customer)
```

## 4. Loop through a dictionary

```python
customer = {
    "customer_id": "C104",
    "visits": 8,
    "status": "active"
}

for key, value in customer.items():
    print(key, "->", value)
```

## 5. Count categories with a dictionary

```python
channels = ["web", "email", "web", "store", "email", "web"]

counts = {}

for channel in channels:
    counts[channel] = counts.get(channel, 0) + 1

print(counts)
```

## 6. Represent records with dictionaries

```python
orders = [
    {"order_id": 101, "channel": "web", "amount": 45},
    {"order_id": 102, "channel": "email", "amount": 72},
    {"order_id": 103, "channel": "web", "amount": 38}
]

for order in orders:
    print(order["order_id"], order["amount"])
```

## 7. Nested dictionaries

```python
customers = {
    "C1": {"visits": 5, "segment": "consumer"},
    "C2": {"visits": 9, "segment": "business"}
}

print(customers["C2"]["segment"])
```

## 8. Create a set

```python
channels = {"web", "email", "web", "store"}

print(channels)
print("Count:", len(channels))
```

## 9. Remove duplicates with a set

```python
raw_channels = ["web", "email", "web", "store", "email"]

unique_channels = set(raw_channels)

print(unique_channels)
```

## 10. Set operations

```python
active = {"C1", "C2", "C3", "C4"}
subscribers = {"C2", "C4", "C5"}

print("Both:", active & subscribers)
print("Either:", active | subscribers)
print("Active only:", active - subscribers)
```

## 11. Membership testing

```python
valid_statuses = {"complete", "cancelled", "pending"}

status = "complete"

print(status in valid_statuses)
```

## 12. Dictionary + set workflow

```python
orders = [
    {"id": 1, "channel": "web"},
    {"id": 2, "channel": "email"},
    {"id": 3, "channel": "web"},
    {"id": 4, "channel": "store"}
]

counts = {}
unique_channels = set()

for order in orders:
    channel = order["channel"]
    counts[channel] = counts.get(channel, 0) + 1
    unique_channels.add(channel)

print("Counts:", counts)
print("Unique:", unique_channels)
```

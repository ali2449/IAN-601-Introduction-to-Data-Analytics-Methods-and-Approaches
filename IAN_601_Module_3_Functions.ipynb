Functions turn repeated steps into named, reusable analytical tasks.

## 1. Define and call a function

```python
def greet():
    print("Hello from IAN 601")

greet()
```

## 2. Put a calculation inside a function

```python
def calculate_profit():
    revenue = 18000
    costs = 12500
    print(revenue - costs)

calculate_profit()
```

## 3. Parameters receive values

```python
def greet_student(name):
    print(f"Welcome, {name}")

greet_student("Jordan")
```

## 4. Pass two analytical inputs

```python
def profit(revenue, costs):
    return revenue - costs

result = profit(18000, 12500)
print(result)
```

## 5. Default parameters

```python
def total_with_tax(amount, tax_rate=0.07):
    return amount * (1 + tax_rate)

print(total_with_tax(100))
print(total_with_tax(100, 0.05))
```

## 6. return is different from print

`print()` displays a value. `return` sends a value back so other code can reuse it.

```python
def margin(revenue, costs):
    profit = revenue - costs
    return profit / revenue

m = margin(18000, 12500)
print(f"Margin: {m:.1%}")
```

## 7. Return more than one value

```python
def order_summary(amounts):
    total = sum(amounts)
    average = total / len(amounts)
    return total, average

total, average = order_summary([45, 72, 91])

print(total)
print(average)
```

## 8. Local variables

```python
def calculate_total(price, quantity):
    subtotal = price * quantity
    return subtotal

print(calculate_total(19.95, 3))

# subtotal exists only inside the function.
```

## 9. Catch a conversion error

```python
def to_number(value):
    try:
        return float(value)
    except ValueError:
        return None

print(to_number("42.5"))
print(to_number("N/A"))
```

## 10. Docstrings

```python
def conversion_rate(visits, signups):
    """Return signup conversion rate as a decimal."""
    if visits == 0:
        return 0
    return signups / visits

print(conversion_rate(250, 37))
help(conversion_rate)
```

## 11. Compose small functions

```python
def clean_status(value):
    return value.strip().lower()

def is_complete(value):
    return clean_status(value) == "complete"

print(is_complete(" Complete "))
print(is_complete("Cancelled"))
```

## 12. Applied example: classify a temperature

```python
def classify_temperature(value):
    if value < 36.0:
        return "low"
    elif value <= 37.5:
        return "normal"
    else:
        return "high"

for temp in [35.8, 36.8, 38.2]:
    print(temp, classify_temperature(temp))
```

## 13. Mini project: reusable calculator

```python
def sales_summary(revenue, costs, customers):
    profit = revenue - costs
    margin = profit / revenue if revenue else 0
    revenue_per_customer = revenue / customers if customers else 0

    return {
        "profit": profit,
        "margin": margin,
        "revenue_per_customer": revenue_per_customer
    }

summary = sales_summary(18000, 12500, 240)
print(summary)
```

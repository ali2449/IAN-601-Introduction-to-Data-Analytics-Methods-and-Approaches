## 1. List comprehensions

```python
sales = [120, 80, 145, 60, 200]

high_sales = [value for value in sales if value >= 100]

print(high_sales)
```

A comprehension combines a loop, an expression, and an optional condition.

```python
values = [1, 2, 3, 4, 5]

squares = [value ** 2 for value in values]

print(squares)
```

## 2. Lambda functions

A lambda is useful for a small one-expression function.

```python
double = lambda x: x * 2

print(double(12))
```

Lambdas are often useful as temporary sorting keys.

```python
orders = [
    {"id": 1, "amount": 45},
    {"id": 2, "amount": 120},
    {"id": 3, "amount": 72}
]

sorted_orders = sorted(
    orders,
    key=lambda order: order["amount"],
    reverse=True
)

print(sorted_orders)
```

## 3. Generators and yield

A generator produces values one at a time instead of building the entire result immediately.

```python
def even_numbers(limit):
    for number in range(limit):
        if number % 2 == 0:
            yield number

for number in even_numbers(10):
    print(number)
```

## 4. Inspect a generator with next()

```python
generator = even_numbers(10)

print(next(generator))
print(next(generator))
print(next(generator))
```

## 5. Debugging techniques

Start with small checks: inspect values, types, lengths, and assumptions.

```python
raw_values = ["12", "18", "N/A", "25"]

for value in raw_values:
    print("Value:", value, "| Type:", type(value))
```

## 6. Use assertions to check assumptions

```python
scores = [88, 92, 79, 95]

assert len(scores) > 0
assert all(score >= 0 for score in scores)

print("Checks passed")
```

## 7. Efficient built-ins

```python
values = [12, 18, 25, 31, 40]

print("Total:", sum(values))
print("Minimum:", min(values))
print("Maximum:", max(values))
print("Any above 35?", any(v > 35 for v in values))
print("All positive?", all(v > 0 for v in values))
```

## 8. Measure performance with perf_counter()

```python
from time import perf_counter

start = perf_counter()

total = sum(range(1_000_000))

elapsed = perf_counter() - start

print("Total:", total)
print("Elapsed seconds:", round(elapsed, 6))
```

## 9. Filter adult patient ages

```python
ages = [12, 18, 25, 44, 16, 67, 31]

adult_ages = [age for age in ages if age >= 18]

print(adult_ages)
```

## 10. Mini analytical generator pipeline

```python
def valid_amounts(values):
    for value in values:
        if value is None:
            continue
        if value < 0:
            continue
        yield value

raw_amounts = [45, None, 72, -5, 91, 120]

clean_amounts = list(valid_amounts(raw_amounts))

print(clean_amounts)
print("Average:", sum(clean_amounts) / len(clean_amounts))
```
